### Errors are normal

Python is very strict: one missing quote or bracket and it refuses to run. That is not a disaster, it is information.
When you get an error, **read the last line first**: it tells you the type of error and usually what is wrong.

Run the cell below. It fails on purpose.

````python
print("Wavely has", 800 users)
````

The last line says `SyntaxError`: Python can't understand the sentence. `users` should be inside the quotes, or removed.
Errors you will meet often:

| Error | Means | Typical cause |
|---|---|---|
| `SyntaxError` | Python can't read the code | missing quote `"`, bracket `)` or colon `:` |
| `NameError` | this name doesn't exist | a typo (`Print`, `pritn`) or a variable you haven't created yet |
| `TypeError` | wrong kind of value | adding text and a number: `"plays: " + 5` |
| `IndentationError` | the spaces at the start of a line are wrong | see Part 5 |

## Part 2 — Numbers and variables

Python is a very good calculator.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | plus, minus, times | `3 * 4` | `12` |
| `/` | divide (always gives a decimal) | `7 / 2` | `3.5` |
| `//` | whole-number divide | `7 // 2` | `3` |
| `%` | remainder ("what is left over") | `7 % 2` | `1` |
| `**` | power | `2 ** 10` | `1024` |

````python
print(10 + 5)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(round(3.14159, 2))  # round to 2 decimals
````

A **variable** is a name for a value. You create it with `=`: the name on the left, the value on the right.
After that, you can use the name instead of the value.

````python
song_seconds = 214
times_played = 15

total_seconds = song_seconds * times_played
print(total_seconds)
````

`=` does **not** mean "is equal" like in maths. It means: *calculate the right side, then store the result under the name on the left*.
That's why this works:

````python
plays = 10
plays = plays + 1  # take the old value (10), add 1, store it again under "plays"
print(plays)
````

**Variable names:** lower case, words joined with `_` (`song_seconds`), no spaces, must not start with a digit.
Upper and lower case matter: `plays` and `Plays` are two different variables.

## Part 3 — Text (strings)

Text in Python is called a **string** (`str`). It goes inside quotes: `"Dua Lipa"` or `'Dua Lipa'`, both work.
You can glue strings together with `+` and measure their length with `len`.

````python
artist = "Dua Lipa"
title = "Levitating"

print(artist + " - " + title)
print(len(artist))  # the space counts too
print("la" * 3)  # repeat a string
````

If the text itself contains a quote, use the other kind of quote around it:

````python
print("Let's dance")  # ' inside "..."
print('She said "play it again"')  # " inside '...'
````

The characters of a string have **positions, starting at 0** (lists in Part 6 work the same way). `text[0]` is the first character,
`text[-1]` the last, and `text[0:4]` the characters at positions 0, 1, 2 and 3 (the end position is not included).

```
"T0128"
 01234
```

````python
track_id = "T0128"

print(track_id[0])  # first character
print(track_id[-1])  # last character
print(track_id[1:])  # from position 1 to the end
print(int(track_id[1:]) + 1)  # the number part as an int, plus 1
````

Strings have **methods**: small built-in tools you call with a dot. They give you a *new* string and leave the old one unchanged.
Data from apps is often messy, so these are a data engineer's daily tools.

````python
messy = "  dUA lIPA  "

print(messy.strip())  # remove spaces at both ends
print(messy.upper())
print(messy.lower())
print(messy.strip().title())  # methods can be chained: first strip, then title
print(messy.replace("U", "u"))
````

A few more you will use a lot. Some answer a yes/no question (they give a `bool`), `split` cuts a string into a **list** of parts
(a list is a row of values in `[ ]`; you will learn all about lists in Part 6):

````python
file_name = "plays_2026-09-01.csv"

print(file_name.startswith("plays_"))  # True
print(file_name.endswith(".csv"))  # True
print("2026" in file_name)  # True: is this text inside?
print(file_name.count("-"))  # 2
print(file_name.upper())

line = "Dua Lipa,Levitating,203"
parts = line.split(",")  # cut at every comma
print(parts)
print(parts[1])
````

**f-strings** are the easiest way to put values into text: write `f` before the quotes and put variables in `{ }`.
The cell below uses `title` and `artist` from the first cell of Part 3.

````python
plays = 1520
print(f"{title} by {artist} was played {plays} times")
print(f"That is {plays * 3.5 / 60} hours")
print(f"That is {plays * 3.5 / 60:.1f} hours")  # :.1f = show 1 decimal
````

### Formatting numbers in f-strings

After the value, a colon `:` starts a **format**: how the value should look. It only changes how the number is *shown*,
not the number itself. The ones you need most:

- `:.1f`, `:.2f` — a decimal number (`f` = float) with exactly 1 or 2 digits after the point
- `:02d`, `:04d` — a whole number (`d` = digits) with at least 2 or 4 digits, filled up with zeros in front
- `:,` — a comma between thousands
- `:.1%` — multiply by 100 and show as a percentage

````python
per_day = 1520 / 7  # average plays per day
print(f"{per_day}")  # without a format: lots of digits
print(f"{per_day:.1f}")  # 1 decimal: 217.1
print(f"{10.99 * 12:.2f}")  # 2 decimals: 131.88
print(f"{5:.2f}")  # also works for whole numbers: 5.00
````

````python
minutes = 3
seconds = 5
print(f"{minutes}:{seconds}")  # 3:5 looks wrong
print(f"{minutes}:{seconds:02d}")  # 3:05

day = 1
month = 9
print(f"{day:02d}.{month:02d}.2026")  # 01.09.2026
print(f"U{42:04d}")  # a user id: U0042
````

````python
print(f"{1520345:,}")  # 1,520,345
print(f"{1520345.678:,.2f}")  # formats can be combined: 1,520,345.68
print(f"{0.0834:.1%}")  # 8.3%
````

Two things that often go wrong:
- `d` only works for whole numbers: `f"{3.5:02d}"` gives a `ValueError`. Use `f` for decimals.
- Formats need a **number**: `f"{'5':.2f}"` fails because `'5'` is text. Convert it first: `f"{float('5'):.2f}"`.

### Strings cheat sheet

| You want | Write | Result |
|---|---|---|
| glue text | `"Dua" + " " + "Lipa"` | `"Dua Lipa"` |
| repeat text | `"la" * 3` | `"lalala"` |
| length | `len("Queen")` | `5` |
| first / last character | `"T0128"[0]`, `"T0128"[-1]` | `"T"`, `"8"` |
| a part | `"T0128"[1:]`, `"T0128"[0:3]` | `"0128"`, `"T01"` |
| remove spaces at both ends | `"  hi  ".strip()` | `"hi"` |
| change case | `.upper()`, `.lower()`, `.title()` | `"DUA"`, `"dua"`, `"Dua Lipa"` |
| replace | `"dj_anna".replace("_", " ")` | `"dj anna"` |
| starts / ends with | `"plays_1.csv".startswith("plays_")`, `"plays_1.csv".endswith(".json")` | `True`, `False` |
| contains | `"20" in "2026"` | `True` |
| count | `"a-b-c".count("-")` | `2` |
| cut into a list | `"a,b,c".split(",")` | `["a", "b", "c"]` |
| text → number | `int("183")`, `float("4.5")` | `183`, `4.5` |
| number → text | `str(183)` | `"183"` |
| value inside text | `f"{artist} ({plays})"` | `"Dua Lipa (1520)"` |

**f-string formats** — `f"{value:FORMAT}"`:

| Format | Means | Example | Result |
|---|---|---|---|
| `.1f` | 1 decimal | `f"{3.14159:.1f}"` | `3.1` |
| `.2f` | 2 decimals | `f"{10.5:.2f}"` | `10.50` |
| `02d` | whole number, at least 2 digits, zeros in front | `f"{5:02d}"` | `05` |
| `04d` | whole number, at least 4 digits | `f"{42:04d}"` | `0042` |
| `,` | thousands separator | `f"{1520345:,}"` | `1,520,345` |
| `,.2f` | both | `f"{1234.5:,.2f}"` | `1,234.50` |
| `.1%` | percentage, 1 decimal | `f"{0.0834:.1%}"` | `8.3%` |
| `>8` / `<8` | right / left aligned in 8 characters | `f"{'AT':>8}"` | `      AT` |

## Part 4 — Types: why `"183"` is not `183`

Every value has a **type**. `type(...)` tells you which one.

| Type | Example | Used for |
|---|---|---|
| `int` | `183` | whole numbers: counts, seconds |
| `float` | `10.99` | decimal numbers: prices, averages |
| `str` | `"183"` | text — always in quotes |
| `bool` | `True` / `False` | yes/no facts (Part 5) |
| `None` | `None` | "no value", missing data |

````python
print(type(183))
print(type(10.99))
print(type("183"))
print(type(True))
print(type(None))
````

This matters a lot for data engineers: **everything read from a file arrives as text**. Text and numbers behave differently:

````python
print("183" + "7")  # two strings: glued together
print(183 + 7)  # two numbers: added
````

Convert between types with `int(...)`, `float(...)` and `str(...)`:

````python
seconds_text = "183"  # as it comes from a file
seconds = int(seconds_text)  # now a number
print(seconds + 7)

price = float("10.99")
print(price * 2)

print("Seconds: " + str(seconds))  # number -> text, so it can be glued to text
````

## Part 5 — True or False, and your first `if`

A **comparison** gives a `bool`: `True` or `False`.

| Operator | Means |
|---|---|
| `==` | equal (two `=`! One `=` stores a value) |
| `!=` | not equal |
| `<` `<=` `>` `>=` | smaller, smaller or equal, bigger, bigger or equal |
| `and` `or` `not` | combine conditions |

````python
seconds = 12
plan = "free"

print(seconds < 30)
print(plan == "premium")
print(plan == "free" and seconds < 30)
print(plan == "premium" or seconds < 30)
````

With `if` the program **decides** what to do. Note the two rules:
- the `if` line ends with a colon `:`
- the lines that belong to it are **indented by 4 spaces** (Colab does this for you when you press Enter after the `:`)

````python
seconds = 12

if seconds < 30:
    print("Skip: this play is not paid to the artist")
else:
    print("This play counts")

print("This line always runs: it is not indented")
````

## Part 6 — Lists and your first loop

A **list** holds many values in order, inside square brackets `[ ]`. Positions are counted **from 0**.

```
countries = ["AT", "SK", "HU", "DE"]
position:      0     1     2     3
from the end: -4    -3    -2    -1
```

````python
countries = ["AT", "SK", "HU", "DE"]

print(len(countries))  # how many values
print(countries[0])  # the first one
print(countries[-1])  # the last one
print(countries[0:2])  # positions 0 and 1 (2 is not included)
print("HU" in countries)  # is it in the list?

countries.append("PL")  # add a value at the end
print(countries)
````

Lists of numbers can be summed and sorted:

````python
song_seconds = [214, 187, 243]

print(sum(song_seconds), max(song_seconds), min(song_seconds))
print(sorted(song_seconds))
print(sorted(song_seconds, reverse=True))  # biggest first
````

### Your first loop

A `for` loop runs the indented lines **once for every value** in a list. Each round, the variable (here `artist`) holds the next value.

````python
artists = ["Dua Lipa", "Coldplay", "Queen"]

for artist in artists:
    print("Now playing:", artist)

print("Playlist finished")  # not indented: runs once, after the loop
````

A very common pattern is the **running total**: start at 0 and add something in every round.

````python
plays_per_day = [120, 95, 143]

total = 0
for plays in plays_per_day:
    total = total + plays
    print("added", plays, "-> total is now", total)

print("Total:", total)
````

## Part 7 — Dictionaries

A **dictionary** (`dict`) stores **key → value** pairs inside curly brackets `{ }`. You look things up by key, not by position,
just like looking up a word in a real dictionary. One dict is perfect for **one row** of data, e.g. one user.

````python
user = {
    "user_id": "U0101",
    "name": "Anna Novak",
    "country": "SK",
    "plan": "free",
}

print(user["name"])  # read a value
user["country"] = "AT"  # change a value (Anna moved to Vienna)
user["age"] = 27  # add a new key
print(user)
print(len(user), "keys")
````

Asking for a key that doesn't exist gives a `KeyError`. Real data often has missing fields, so use `.get(key, default)`:
it returns the default instead of crashing.

````python
print(user.get("city"))  # None, no crash
print(user.get("city", "unknown"))  # with a default
````

### A list of dictionaries is a table

Put many dicts in a list, and you have a **table**: every dict is a row, every key a column.
This is the most important idea for session 1, and the way data engineers think about data.

| artist | seconds |
|---|---|
| Dua Lipa | 214 |
| Coldplay | 243 |
| Queen | 354 |

````python
plays = [
    {"artist": "Dua Lipa", "seconds": 214},
    {"artist": "Coldplay", "seconds": 243},
    {"artist": "Queen", "seconds": 354},
]

print(plays[2])  # the third row
print(plays[2]["artist"])  # third row, column "artist"

for play in plays:
    print(play["artist"], "played for", play["seconds"], "seconds")
````

### Exercise 10 (Stretch) — Your Wavely Wrapped
Everything from today together. Using `plays` below:
1. `total_minutes` — total listening time in minutes, rounded to 1 decimal
2. `dua_lipa_plays` — how many plays are by Dua Lipa
3. `longest_artist` — the artist of the longest play. Hint: remember the longest seconds so far in a variable and update it in the loop

````python
plays = [
    {"artist": "Dua Lipa", "seconds": 214},
    {"artist": "Coldplay", "seconds": 243},
    {"artist": "Dua Lipa", "seconds": 187},
    {"artist": "Billie Eilish", "seconds": 12},
    {"artist": "Queen", "seconds": 354},
]

total_seconds = 0
dua_lipa_plays = 0
longest_seconds = 0
longest_artist = None
for play in plays:
    total_seconds = total_seconds + play["seconds"]
    if play["artist"] == "Dua Lipa":
        dua_lipa_plays = dua_lipa_plays + 1
    if play["seconds"] > longest_seconds:
        longest_seconds = play["seconds"]
        longest_artist = play["artist"]
total_minutes = round(total_seconds / 60, 1)

print(f"You listened for {total_minutes} minutes")  # expected: You listened for 16.8 minutes
print(f"Dua Lipa plays: {dua_lipa_plays}")  # expected: Dua Lipa plays: 2
print(f"Your longest play: {longest_artist}")  # expected: Your longest play: Queen
````

### Practice 2 — Count the skips
Plays shorter than 30 seconds are **skips**. Count them in `plays_seconds`, and also compute `share`: the skips as a percentage of all plays, rounded to 1 decimal.

````python
plays_seconds = [12, 200, 45, 8, 310, 29, 187]

skips = 0
for s in plays_seconds:
    if s < 30:
        skips = skips + 1
share = round(skips / len(plays_seconds) * 100, 1)

print(skips, share)  # expected: 3 42.9
````

### Practice 3 — Country names
Wavely stores countries as two-letter codes. Loop over `codes` and print the full name of each country, using the dict `country_names`.

````python
country_names = {"AT": "Austria", "SK": "Slovakia", "HU": "Hungary", "CZ": "Czechia", "DE": "Germany", "PL": "Poland"}
codes = ["HU", "AT", "PL"]

for code in codes:
    print(country_names[code])

# expected:
# Hungary
# Austria
# Poland
````
