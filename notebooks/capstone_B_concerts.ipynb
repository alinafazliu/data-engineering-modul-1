Wavely started selling concert tickets in the app. The ticketing partner sends one big, messy export.
The live-events team asks:
1. How much **revenue** did each concert (and each artist) make so far?
2. How full are the venues? (**sell-through** = tickets sold / venue capacity)
3. How do sales split by **ticket type**, and how early before the concert do people buy?

Build a small, trustworthy pipeline that answers these questions and present it in session 10.

## The data (`data/capstone/concerts/`)
| File | Columns |
|---|---|
| `venues.csv` | `venue_id, venue_name, city, country, capacity` |

| `ticket_sales_messy.csv` | `sale_id, sold_at, concert_id, ticket_type, quantity, price, buyer_country` |

`price` is the price **per ticket** in EUR. The export is messy in several ways — finding out how is part of the job.
