### Errors are normal

Python is very strict: one missing quote or bracket and it refuses to run. That is not a disaster, it is information.
When you get an error, **read the last line first**: it tells you the type of error and usually what is wrong.

Run the cell below. It fails on purpose.

````python
print("Wavely has", 800 users)
````

The last line says `SyntaxError`: Python can't understand the sentence. `users` should be inside the quotes, or removed.
Errors you will meet often:

| Error | Means | Typical cause |
|---|---|---|
| `SyntaxError` | Python can't read the code | missing quote `"`, bracket `)` or colon `:` |
| `NameError` | this name doesn't exist | a typo (`Print`, `pritn`) or a variable you haven't created yet |
| `TypeError` | wrong kind of value | adding text and a number: `"plays: " + 5` |
| `IndentationError` | the spaces at the start of a line are wrong | see Part 5 |

## Part 2 — Numbers and variables

Python is a very good calculator.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | plus, minus, times | `3 * 4` | `12` |
| `/` | divide (always gives a decimal) | `7 / 2` | `3.5` |
| `//` | whole-number divide | `7 // 2` | `3` |
| `%` | remainder ("what is left over") | `7 % 2` | `1` |
| `**` | power | `2 ** 10` | `1024` |

````python
print(10 + 5)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(round(3.14159, 2))  # round to 2 decimals
````

A **variable** is a name for a value. You create it with `=`: the name on the left, the value on the right.
After that, you can use the name instead of the value.

````python
song_seconds = 214
times_played = 15

total_seconds = song_seconds * times_played
print(total_seconds)
````

`=` does **not** mean "is equal" like in maths. It means: *calculate the right side, then store the result under the name on the left*.
That's why this works:

````python
plays = 10
plays = plays + 1  # take the old value (10), add 1, store it again under "plays"
print(plays)
````

**Variable names:** lower case, words joined with `_` (`song_seconds`), no spaces, must not start with a digit.
Upper and lower case matter: `plays` and `Plays` are two different variables.

## Part 3 — Text (strings)

Text in Python is called a **string** (`str`). It goes inside quotes: `"Dua Lipa"` or `'Dua Lipa'`, both work.
You can glue strings together with `+` and measure their length with `len`.

````python
artist = "Dua Lipa"
title = "Levitating"

print(artist + " - " + title)
print(len(artist))  # the space counts too
````

Strings have **methods**: small built-in tools you call with a dot. They give you a *new* string and leave the old one unchanged.
Data from apps is often messy, so these are a data engineer's daily tools.

````python
messy = "  dUA lIPA  "

print(messy.strip())  # remove spaces at both ends
print(messy.upper())
print(messy.lower())
print(messy.strip().title())  # methods can be chained: first strip, then title
print(messy.replace("U", "u"))
````

**f-strings** are the easiest way to put values into text: write `f` before the quotes and put variables in `{ }`.

````python
plays = 1520
print(f"{title} by {artist} was played {plays} times")
print(f"That is {plays * 3.5 / 60} hours")
print(f"That is {plays * 3.5 / 60:.1f} hours")  # :.1f = show 1 decimal
````

## Part 4 — Types: why `"183"` is not `183`

Every value has a **type**. `type(...)` tells you which one.

| Type | Example | Used for |
|---|---|---|
| `int` | `183` | whole numbers: counts, seconds |
| `float` | `10.99` | decimal numbers: prices, averages |
| `str` | `"183"` | text — always in quotes |
| `bool` | `True` / `False` | yes/no facts (Part 5) |
| `None` | `None` | "no value", missing data |

````python
print(type(183))
print(type(10.99))
print(type("183"))
print(type(True))
print(type(None))
````

This matters a lot for data engineers: **everything read from a file arrives as text**. Text and numbers behave differently:

````python
print("183" + "7")  # two strings: glued together
print(183 + 7)  # two numbers: added
````

Convert between types with `int(...)`, `float(...)` and `str(...)`:

````python
seconds_text = "183"  # as it comes from a file
seconds = int(seconds_text)  # now a number
print(seconds + 7)

price = float("10.99")
print(price * 2)

print("Seconds: " + str(seconds))  # number -> text, so it can be glued to text
````

## Part 5 — True or False, and your first `if`

A **comparison** gives a `bool`: `True` or `False`.

| Operator | Means |
|---|---|
| `==` | equal (two `=`! One `=` stores a value) |
| `!=` | not equal |
| `<` `<=` `>` `>=` | smaller, smaller or equal, bigger, bigger or equal |
| `and` `or` `not` | combine conditions |

````python
seconds = 12
plan = "free"

print(seconds < 30)
print(plan == "premium")
print(plan == "free" and seconds < 30)
print(plan == "premium" or seconds < 30)
````

With `if` the program **decides** what to do. Note the two rules:
- the `if` line ends with a colon `:`
- the lines that belong to it are **indented by 4 spaces** (Colab does this for you when you press Enter after the `:`)

````python
seconds = 12

if seconds < 30:
    print("Skip: this play is not paid to the artist")
else:
    print("This play counts")

print("This line always runs: it is not indented")
````

## Part 6 — Lists and your first loop

A **list** holds many values in order, inside square brackets `[ ]`. Positions are counted **from 0**.

```
countries = ["AT", "SK", "HU", "DE"]
position:      0     1     2     3
from the end: -4    -3    -2    -1
```

````python
countries = ["AT", "SK", "HU", "DE"]

print(len(countries))  # how many values
print(countries[0])  # the first one
print(countries[-1])  # the last one
print(countries[0:2])  # positions 0 and 1 (2 is not included)
print("HU" in countries)  # is it in the list?

countries.append("PL")  # add a value at the end
print(countries)
````

Lists of numbers can be summed and sorted:

````python
song_seconds = [214, 187, 243]

print(sum(song_seconds), max(song_seconds), min(song_seconds))
print(sorted(song_seconds))
print(sorted(song_seconds, reverse=True))  # biggest first
````

### Your first loop

A `for` loop runs the indented lines **once for every value** in a list. Each round, the variable (here `artist`) holds the next value.

````python
artists = ["Dua Lipa", "Coldplay", "Queen"]

for artist in artists:
    print("Now playing:", artist)

print("Playlist finished")  # not indented: runs once, after the loop
````

A very common pattern is the **running total**: start at 0 and add something in every round.

````python
plays_per_day = [120, 95, 143]

total = 0
for plays in plays_per_day:
    total = total + plays
    print("added", plays, "-> total is now", total)

print("Total:", total)
````

## Part 7 — Dictionaries

A **dictionary** (`dict`) stores **key → value** pairs inside curly brackets `{ }`. You look things up by key, not by position,
just like looking up a word in a real dictionary. One dict is perfect for **one row** of data, e.g. one user.

````python
user = {
    "user_id": "U0101",
    "name": "Anna Novak",
    "country": "SK",
    "plan": "free",
}

print(user["name"])  # read a value
user["country"] = "AT"  # change a value (Anna moved to Vienna)
user["age"] = 27  # add a new key
print(user)
print(len(user), "keys")
````

Asking for a key that doesn't exist gives a `KeyError`. Real data often has missing fields, so use `.get(key, default)`:
it returns the default instead of crashing.

````python
print(user.get("city"))  # None, no crash
print(user.get("city", "unknown"))  # with a default
````

### A list of dictionaries is a table

Put many dicts in a list, and you have a **table**: every dict is a row, every key a column.
This is the most important idea for session 1, and the way data engineers think about data.

| artist | seconds |
|---|---|
| Dua Lipa | 214 |
| Coldplay | 243 |
| Queen | 354 |

````python
plays = [
    {"artist": "Dua Lipa", "seconds": 214},
    {"artist": "Coldplay", "seconds": 243},
    {"artist": "Queen", "seconds": 354},
]

print(plays[2])  # the third row
print(plays[2]["artist"])  # third row, column "artist"

for play in plays:
    print(play["artist"], "played for", play["seconds"], "seconds")
````
